### Import libraries

```python
from pathlib import Path

from resmip import Image, RTStructureSet
```

## READ DICOM AND SAVE NII 

### Define path

```python
dicom_path = Path.cwd().parent / "tests" / "Dicom" / "IBSI1_CT_phantom"
outputs = Path.cwd().parent / "src" /"wrk"
dicom_path, outputs
```

### Read original dicom image and dicom rtstruct

```python
# Read original dicom image
image_ct = Image.read(dicom_path / "CT_00000")
# Read original dicom rtstruct
path_rt = list((dicom_path).glob("RTst_*/*.dcm"))[0]
structures = RTStructureSet.read(path_rt, reference_image = image_ct)
print(structures)
```

### Define path where to save nii files

```python
save_path = outputs / "prove"
save_path.mkdir(exist_ok=True, parents=True)
```

### Save nifti (image and rt structure)

```python
image_ct.write(save_path / "img.nii.gz", write_metadata = True)
structures["GTV-1"].write(save_path, file_format = ".nii.gz")
```

## OPPOSITE PROCESS (READ NII AND SAVE DICOM): 

### Read nifti files (image and rt structure)

```python
new_image = Image.read(save_path / "img.nii.gz")
new_structure = RTStructureSet.read([save_path / "GTV-1.nii.gz"])
```

### Save path where dicom CT and structeures are saved

```python
save_path_dcm = save_path / "dicom"
save_path_dcm.mkdir(exist_ok=True, parents=True)
```

### Write dicom image

```python
new_image.write(save_path_dcm)
```

### Write dicom RTStruct

```python
new_structure.write(save_path_dcm / "RS.dcm", reference_image_path = save_path_dcm)
```
